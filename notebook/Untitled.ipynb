```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
df = pd.read_csv("SampleSuperstore.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.drop_duplicates(inplace=True)
```

```python
top_subcategory = df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False)
```

```python
top_subcategory
```

```python
top_subcategory.plot(kind='bar', figsize=(10,5))
plt.title("Top Selling Sub-Categories")
plt.show()
```

```python
top_category = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

top_category.plot(kind='bar')
plt.title("Top Selling Categories")
plt.show()
```

```python
state_sales = df.groupby("State")["Sales"].sum().sort_values(ascending=False)

state_sales.head(10).plot(kind="bar")
plt.title("Top States by Sales")
plt.show()
```

```python
city_sales = df.groupby("City")["Sales"].sum().sort_values(ascending=False)

city_sales.head(10).plot(kind="bar", figsize=(10,5))
plt.title("Top Cities by Sales")
plt.show()
```

```python
segment_sales = df.groupby("Segment")["Sales"].sum()

segment_sales.plot(kind="pie", autopct="%1.1f%%")
plt.title("Sales by Segment")
plt.show()
```

```python
profit_sub = df.groupby("Sub-Category")["Profit"].sum().sort_values(ascending=False)

profit_sub.plot(kind="bar")
plt.title("Profit by Sub-Category")
plt.show()
```

```python
sns.scatterplot(data=df, x="Discount", y="Profit")
plt.title("Discount vs Profit")
plt.show()
```

```python
sns.scatterplot(data=df, x="Quantity", y="Sales")
plt.show()
```
